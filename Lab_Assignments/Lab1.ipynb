# Matplotlib Tasks

```python
import matplotlib.pyplot as plt
```

T-1

```python
temperature_dhaka =np.array([25,34,21,45,28,6,43,18,7,2])
humidity_dhaka =np.array([28, 25,29,20, 26, 50, 19, 29, 52, 55])
plt.xlabel("temperature_dhaka")
plt.ylabel("humidity_dhaka")
plt.title("temperature_dhaka vs humidity_dhaka")
plt.scatter(temperature_dhaka, humidity_dhaka, marker ="*",color ="r")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

subjects = np.array(['Maths', 'English', 'Science', 'Physics', 'Computer'])
marks = np.array([89, 90, 45, 78, 99])
colors = ["r","g","b","y","pink"]
plt.figure(figsize=(10, 6))
plt.bar(subjects,marks, color= colors)
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Scores in Each Subject")
plt.ylim(0, 100)
plt.grid(axis="y")
plt.show()
```

T-5

```python
temperature_dhaka = np.array([25,34,21,45,28,6,43,18,7,2])
humidity_dhaka = np.array([28, 25,29,20, 26, 50, 19, 29, 52, 55])


study_hours = np.array([2,3,4,4, 5, 6, 7, 7, 8, 9, 9, 10, 11, 11, 12])
marks_for_study_hours  = np.array([6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100])


subjects = np.array(['Maths', 'English', 'Science', 'Physics', 'Computer'])
marks_for_subjects = np.array([89, 90, 45, 78, 99])
colors = ["r","g","b","y","pink"]

fig = plt.figure(figsize=(15,10))

y1 = fig.add_subplot(2,2,1)
y1.scatter(temperature_dhaka, humidity_dhaka,marker = "*",color ="r")
y1.set_xlabel("Temperature")
y1.set_ylabel("Humidity")
y1.set_title("temperature_dhaka vs humidity_dhaka")

y2 = fig.add_subplot(2,2,2)
y2.plot(study_hours, marks_for_study_hours, color ="g")
y2.set_xlabel("Study Hours")
y2.set_ylabel("Marks")
y2.set_title("study_hours vs marks")

y3 = fig.add_subplot(2,2,3)
y3.barh(subjects, marks_for_subjects)
y3.set_xlabel("Marks")
y3.set_ylabel("Subjects")
y3.set_title("Scores in Each Subject(Horizontal)")
y3.set_xlim(0,100)
y3.grid(axis="x")


y4 = fig.add_subplot(2, 2, 4)
y4.bar(subjects,marks_for_subjects, color= colors)
y4.set_xlabel("Subjects")
y4.set_ylabel("Marks")
y4.set_title("Scores in Each Subject (Vertical)")
y4.set_ylim(0,100)
y4.grid(axis="y")
plt.show()
```
